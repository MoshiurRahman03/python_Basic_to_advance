```python
#write a function takes a List as input and returns  their average

def  avg_list(lst) :
    result=0
    count=0
    for i in  lst:
        result=result+i
        count=count+1
    avg=result/count    
    print("Average of the numbers in the list is:",avg)
  
lst=[1,2,3,4,5,6]
avg_list(lst)
```

```python
# POP() is function shows  the any member and removes 

fruits=["Apple","Mango","Banana","Orange"]
num=input("Enter the number you want to pop up and remove:")
num=int(num)
ln=len(fruits)
for i in range(ln) :
    if num>ln :
        print(" pop index out of range!")
        break
    else :
        print(" pop index in the range!")
        break
        
print("Before popping:",fruits)
item=fruits.pop(num)
print(item)
print("After popping",fruits)
```
