```python
#basic operation
str1="This is a string."
str2='This is another string.'
str3="""This is another string"""
str4="This is a string. \n we are crrating it in python" ; print(str4)
str5="This is a string. \t we are crrating it in python" ; print(str5)

#concatenate
str12=str1+str2 ; print(str12) ; print(len(str12))
```

```python
str="My college"
#indexing: position of character in a index
for i in range(len(str)):
  print(str[i])
```

```python
#slicing 
#str[starting_index : ending_index]
str="My college" ; print(str[0:4]) ;print(str[:4])
print(str[4:len(str)])
print(str[4:])
```

```python
#negetive indexing 
str="apple"
print(str[-5:-2]) ;print(str[:-2]) ;print(str[-5:])
```

```python
#some string functions
str="Bangladeshi python coder"
print(str.endswith("er"))
print(str.capitalize()) ;print(str)  #capitalize 1st alphabet
print(str.replace("python","java"))
print(str.find("a"))  #index of the poition where 1st occured
print("How many time O occured:" ,str.count("o"))
```
