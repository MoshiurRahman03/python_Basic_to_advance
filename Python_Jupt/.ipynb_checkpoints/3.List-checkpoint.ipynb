```python
#list is built in data type that stores set of values(integer, float, string etc)
marks=[94.4,87.5, 95.2, 66.4, 45.1]
print(marks) ; print(type(marks))
for i in range(len(marks)):
    print(marks[i])
```

```python
#list with various data
student=["Rafiq",95 ,17,"Dhaka"] 
for i in range(len(student)):
    print(student[i])
student[0]="Karim"
print("\n new updated list")
for i in range(len(student)):
    print(student[i])
```

```python
#List slicing
marks=[87,64,33,95,76]
print(marks)
print(marks[1:4]) 
print(marks[:4])   #0th index theke4th index 
print(marks[1:])   #1st index theke baki gula
print(marks[-3:-1])
```

```python
#list specific method
list=[2,1,3,1,4] ;print(list)
list.append(4) ; print(list)
list.sort() ; print(list)   #ascending order
list.sort(reverse=True) ; print(list) #descending order
#insert element at index
list.insert(1,5) ; print(list)  #installed 5 at position 1
list.remove(5) ; print(list)   # 5 jotobar ase, tar 1st occurance withdraw korbe
list.pop(3) ; print(list)  #certain index er element ke pop korbe
```
