```python
#tuple is built in data type which is immutable sequence that can't be changed
tup=(2,3,4,1)
print(type(tup))
for i in range(len(tup)):
    print(tup[i])

tup1=("hello") ; print(type(tup1))  #it is string, so, to be tuple there should be a comma 
tup2=("hello",) ; print(type(tup2))

print(tup[1:3]) ;print(tup[0:]); print(tup[:3]); print(tup[-4:-1])

print(tup.index(2) ,tup.count(3))
```
