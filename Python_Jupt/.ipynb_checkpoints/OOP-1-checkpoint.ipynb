```python
# Class is a collection of 2 things one is methods and another is class
# Methods are function that belongs to objects
class student:
    
    college_name="NotreDame College" #clss attribute 

    
    def __init__(self,name,marks):
        self.name=name  #object attributr
        self.marks=marks

    def welcome(self):
         print("Welcome Dear Student,",self.name)
    
    def get_marks(self):
        return self.marks
        

s1=student("Moshiur", 99) ; print(s1.name, s1.marks)
s1.welcome()
s1.get_marks()
```

```python
#static method
# Methods that don't use the self parameter(work at class level)
#decorator allows us to wrap another function in order to extend the behaviour of the wrapped function,without permanently modify it.
class student:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks
    @staticmethod #static method is a decorator 
    def hello():
        print("Herllo World")
    
s1=student("moshiur",25)
s1.hello()
```

```python
# 4 pillers in oop. Abstraction,Encapsulation,Inheritance,potymorphism
#Abstraction:Hiding the implementation details of a class and showing the essential features to the user
class car:
    def __init__ (self):
        self.acc= False
        self.brk= False
        self.clutch= False

    def start(self):
        self.acc= True
        self.clutch= True
        print("car started.....")
        

car1=car()
car1.start()
#unnecessary details have been hiden 
```

```python
#Encapsulation : wrapping data and function into a single unit(object)
class account:
    def __init__(self,bal,acc):
        self.balance=bal
        self.account_name=acc
    #debit method
    def debt(self,amount):
        self.balance=self.balance -amount
        print("Rs.", amount,"has been debited")
        print("total balance:",self.balance)

    #credit method
    def credit(self,amount):
        self.balance=self.balance+amount
        print("Rs.", amount,"has been credited")
        print("total balance:",self.balance)
    
    def get_balance(self):
        return self.balance

acc1=account(1000,12345)  
print(acc1.balance, acc1.account_name)
acc1.credit(200)
acc1.debt(300)
```
