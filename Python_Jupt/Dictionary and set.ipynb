```python
#Dictionary in python
#data value is stored in key:value pair
# "key" :value
dict ={
     "name":"moshiur",   
     "AGE":20,
     "Is adult" :True ,
     "marks":[10, 15 ,20] ,
    "subjects":["Physics","Chemistry","Math"] ,
     "Institution":"BUET" ,
     "Topics" :("dict","set")
}

print(dict)
print(type(dict))

#Accesing Dictionary
print(dict["name"])
print(dict["marks"])
print(dict["subjects"])

#Assigning new value
dict["name"]="Rahman"  #overwrite the old value
print(dict["name"])
```

```python
null_dict={}
print(null_dict)
null_dict["name"]="moshiur"
print(null_dict["name"])
print(null_dict)
```

```python
#Nested dictionary
student={
    "name":"moshiur" ,
    "subjects" :{
        "physics":96,       
        "chemistry":91,       
        "math":100,       
       
                }

}
print(student)
print(student["subjects"])
print(student["subjects"]["math"])
```

```python
#methods in dictonary
student={
    "name":"moshiur" ,
    "subjects" :{
        "physics":96,       
        "chemistry":91,       
        "math":100,       
       
                }

}
print(student.keys())
print(list(student.keys()))
print(len(student))
print(student.values( )) #returns all values
print(student.items( )) #returns all (key, val) pairs as tuples
print(student.get( "key")) 

newDict={"city"}
student.update( newDict )
```
