```python
#del keyword
#used to delete object property and or object itself
class student:
    def __init__ (self,name,mark):
        self.name=name
        self.mark=mark

s1=student("Moshiur",30)
print(s1)
print(s1.name,s1.mark)
del s1.mark 
print(s1.name) 
#print(s1.mark)  #mark attribute has been deleted
del s1 ; print(s1) #object has been deleted
```

```python
#private attribte and method are used only in class but not accessible from outside the class
class Account:
    def __init__(self,acc_no,acc_key_public,acc_key_private):
        
        self.acc_no=acc_no #it is accesible outside of class
        self.acc_key_public=acc_key_public
        self.__acc_key_private=acc_key_private  #it is private ,cant be accessed from outside 
    
    def show_private_key(self):
        print("private key is:",self.__acc_key_private)

    def __private_method(self):
        print("Hello it is a private method") #it is a private method, can't be accesed outside of class

    def access_private_method(self):
        self.__private_method()
       
        
acc1=Account("12345","abcde","1@#$5bn")
print(acc1.acc_no,acc1.acc_key_public)

acc1.show_private_key() #private attribute has been accessed through a function in the class 
acc1.access_private_method()

#acc1.__private_method()
#print(acc1.__acc_key_private)
```

```python
#Inheritances: when a class derives the properties and methods from different class
#single level inheritance (parebt to child)
class car:
    @staticmethod
    def start():
        print("Car started")
    
    
    @staticmethod
    def stop():
        print("Car stopped")


class tyota(car): 
    def __init__ (self,name):
        self.name=name


car1=tyota("fortuner")
car2=tyota("Prius")

print(car1.name,car1.start())
```

```python
#Multilevel inheritance

class car:
    @staticmethod
    def start():
        print("Car started.....")
    
    
    @staticmethod
    def stop():
        print("Car stopped......")


class tyota(car): 
    def __init__ (self,brand):
        self.brand=brand


class fortune(tyota):
    def __init__ (self,type):
        self.type=type


car1=fortune("diesel")
car1.start()
```

```python
#Multiple inheritance
class A:
    varA="Welcome to Class A"

class B:
    varB="Welcome to Class B"

class c(A,B):
    varC="Welcome to Class C"


c1=c()   
print(c1.varC)
print(c1.varA)
print(c1.varB)
```

```python
#sper() methodvis used to accces the method of parent class
class car:
    def __init__(self,type):
        self.type=type

        
    @staticmethod
    def start():
        print("Car started")
    
    
    @staticmethod
    def stop():
        print("Car stopped")


class tyota(car): 
    def __init__ (self,name,type):
        self.name=name
        super().__init__(type)
        super().start

car1=tyota("Prius","electric")
print(car1.type)
```
