```python
#write a function which calculates the average of n numbers
n=int(input("How many numbers:"))
num=[];
def clc_avg_sum():
    smm=0 ;
    for i in range(n):
        value=int(input(f"{i} th number"))
        num.append(value)
    print("Taken inputs are:",num)
    for j in range (len(num)):
        smm=smm+num[j]
    avg= smm/j
    print("The sum and avg of taken inputs are:")
    return smm ,avg 

clc_avg_sum()    
```

```python
# factorial calculating function
def calc_fact():
    n=int(input("Enter n:"))
    fact=1 ;
    for i in range(1,n+1):
        fact=fact*i
    print(f"{n}!=",fact)

calc_fact()    
```

```python
#Recursion
# print n to 1 backward
def show(n):
    #n=int(input("Enter a number:"))
    if(n==0):
        return #if n==0 then it will delete its layer and retuen to presious layer it is called base case
    print(n)
    show(n-1)

show(4);
```

```python
#Recursion through factorial 

def fact_n(n):
    if(n==0 or n==1):
        return 1   
    else:
        return n*fact_n(n-1)

y=int(input("Enter n:"))

print(f"{y}!=" ,fact_n(y))    
```
