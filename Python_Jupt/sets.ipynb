```python
#set is the collection of the unordered items (no index)
#set is also immutable
#set will avoid duplicate value

collection={1,2,3,4,2,"Hello","worlds","Hello" ,4}
print(collection)
print(type(collection))
print(len(collection))
```

```python
coll1={} #it is dictionary
coll2=set() #empty set
print("type of coll1:",type(coll1))
print("type of coll2:",type(coll2))
```

```python
#methods of set
collection=set()
collection.add(1)
collection.add(2)
collection.add(2)
collection.add(3)
collection.add("Hello")
collection.add((1,2,3))
print(collection)

collection.remove(2)
print(collection,"length of the set:" ,len(collection))
collection.clear()  #clear the set
print(collection,"length of the set:" ,len(collection)) 


#pop operation
col={"Hello","worlds","name" ,"Hi"}
print(col.pop())

#union of 2 sets
set1={1,2,3}
set2={3,4,"Hi"}
set3={"Hello","World"}

print(set1.union(set2))
print(set1.intersection(set2))
```
